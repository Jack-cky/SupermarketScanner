```python
import os
from pathlib import Path

import supervision as sv

from smktscnr import SupermarketScanner

os.chdir("..")
```

#### Load SupermarketScanner

```python
model = SupermarketScanner(os.path.join("checkpoints", "smktscnr_ft.onnx"))
```

#### Scan a Snapshot of a Transcation

```python
img_pred, _ = model.scan(
    src=os.path.join("data", "baskets", "sample_basket.jpg"),
    summary=False,
    save=False,
)

sv.plot_image(image=img_pred, size=(8, 8))
```

#### Checkout a Basket

```python
# checkout a transaction video
model.checkout(os.path.join("data", "samples", "sample_transaction.mp4"))
```

```python
# checkout at real time
model.checkout()
```

#### Checkout Sample Baskets

```python
pth_data = Path("data", "baskets")
for file in pth_data.glob("*.jpg"):
    _ = model.scan(str(file), save=True)

pth_data = Path("data", "samples")
for file in pth_data.glob("*.mp4"):
    model.checkout(str(file))

pth_data = Path("data", "transactions")
for file in pth_data.glob("*.mp4"):
    model.checkout(str(file))
```
